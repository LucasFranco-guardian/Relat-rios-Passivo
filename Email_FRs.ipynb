```python
import smtplib
import os
from email.mime.multipart import MIMEMultipart
from email.mime.text import MIMEText
from email.mime.base import MIMEBase
from email import encoders
from datetime import datetime
```

```python
def send_email_with_pdf_FRs(pdf_path):
    # Configurações do remetente e destinatários
    sender_email = 'lc@guardian-asset.com'  # Substitua pelo seu e-mail
    sender_password = os.getenv("senha_gmail")  # Substitua pela sua senha ou senha de aplicativo
    recipient_email = ['lc@guardian-asset.com','pmk@guardian-asset.com', 'msf@guardian-asset.com', 'cm@guardian-asset.com', 'lb@guardian-asset.com', 'ja@guardian-asset.com', 'gsa@guardian-asset.com',
                       'gvm@guardian-asset.com', 'rt@guardian-asset.com', 'acl@guardian-asset.com', 'lt@guardian-asset.com', 'tfa@guardian-asset.com']  # Lista de destinatários

    # Criando a mensagem
    msg = MIMEMultipart()
    msg['From'] = sender_email
    msg['To'] = ", ".join(recipient_email)
    msg['Subject'] = "Fatos Relevantes — Guardian Gestora | 19–20/10/2026"

    # Corpo do e-mail (HTML para preservar negrito, itálico e as setas)
    body = """\
<html>
    <p>Olá, time,</p>

    <p>Resumo dos fatos relevantes do período <b>19–20/10/2026</b>: 5 comunicados — <b>2 positivos e 3 neutros</b>. Relatório completo em anexo.</p>

    <p><b>TRXF11 · TRX Real Estate FII — Positivo</b><br>
    Aprovado o primeiro programa de recompra, de até 9.591.509 cotas (até 10% da classe única), com vigência até 19/10/2027. As cotas recompradas serão canceladas na hora, e o objetivo é gerar valor quando a cota estiver abaixo do valor patrimonial.<br>
    → <i>Geração de valor patrimonial por cota remanescente ao adquirir cotas com desconto sobre o patrimônio líquido.</i></p>

    <p><b>INHF11 · Inter Hedge FII — Positivo</b><br>
    Aprovada a recompra de até 3.340.365 cotas (até 10% do total), de 20/10/2026 a 19/10/2027. A compra será feita abaixo do valor patrimonial do dia anterior, e as cotas adquiridas serão canceladas.<br>
    → <i>Oportunidade de alocação de caixa com ganho patrimonial aos cotistas remanescentes e suporte à liquidez.</i></p>

    <p><b>Sem ticker · Suno Imóveis FII — Neutro</b><br>
    Começou a oferta primária da 1ª emissão: 100.000.000 cotas a R$ 10,00, no total de até R$ 1,0 bilhão. A subscrição abre em 20/10/2026.<br>
    → <i>Lançamento de grande captação para formação de carteira imobiliária.</i></p>

    <p><b>Sem ticker · Mauá Capital Propriedades Logísticas FII — Neutro</b><br>
    Começou a oferta primária da 1ª emissão (rito automático CVM 160): 2.000.000 cotas a R$ 100,00, no total de R$ 200 milhões. A oferta é só para investidores profissionais.<br>
    → <i>Abertura de captação voltada ao segmento logístico com recursos destinados à aquisição de galpões.</i></p>

    <p><b>RBRR11 · RBR High Grade FII — Neutro</b><br>
    O prazo para votar na consulta formal em curso foi estendido até 23/10/2026, a pedido de cotistas.<br>
    → <i>Extensão de prazo para deliberação de cotistas em processo de consulta formal.</i></p>

    <p>Att,<br>

    Lucas Cavalcante.</p>
  </body>
</html>
"""
    msg.attach(MIMEText(body, 'html'))

    # Anexar o arquivo PDF
    try:
        with open(pdf_path, "rb") as attachment:
            part = MIMEBase("application", "octet-stream")
            part.set_payload(attachment.read())

        # Codificar o arquivo em Base64
        encoders.encode_base64(part)

        # Adicionar cabeçalhos para o anexo
        part.add_header(
            "Content-Disposition",
            f"attachment; filename= {os.path.basename(pdf_path)}",  # Nome do arquivo
        )

        # Anexar o arquivo PDF à mensagem
        msg.attach(part)

    except Exception as e:
        print(f"Falha ao anexar o arquivo: {e}")
        return

    # Conectar ao servidor SMTP do Gmail e enviar o e-mail
    try:
        with smtplib.SMTP_SSL('smtp.gmail.com', 465) as smtp_server:
            smtp_server.login(sender_email, sender_password)
            smtp_server.sendmail(sender_email, recipient_email, msg.as_string())

        print(f"E-mail enviado para {', '.join(recipient_email)}")

    except Exception as e:
        print(f"Falha ao enviar e-mail: {e}")
```
