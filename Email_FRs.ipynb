```python
import smtplib
import os
from email.mime.multipart import MIMEMultipart
from email.mime.text import MIMEText
from email.mime.base import MIMEBase
from email import encoders
from datetime import datetime
```

```python
def send_email_with_pdf_FRs(pdf_path):
    # Configurações do remetente e destinatários
    sender_email = 'lc@guardian-asset.com'  # Substitua pelo seu e-mail
    sender_password = os.getenv("senha_gmail")  # Substitua pela sua senha ou senha de aplicativo
    recipient_email = ['lc@guardian-asset.com','pmk@guardian-asset.com', 'msf@guardian-asset.com', 'cm@guardian-asset.com', 'lb@guardian-asset.com', 'ja@guardian-asset.com', 'gsa@guardian-asset.com',
                       'gvm@guardian-asset.com', 'rt@guardian-asset.com', 'acl@guardian-asset.com', 'lt@guardian-asset.com', 'tfa@guardian-asset.com']  # Lista de destinatários

    # Criando a mensagem
    msg = MIMEMultipart()
    msg['From'] = sender_email
    msg['To'] = ", ".join(recipient_email)
    msg['Subject'] = "Fatos Relevantes — Guardian Gestora | 04–05/08/2026"

    # Corpo do e-mail (HTML para preservar negrito, itálico e as setas)
    body = """\
<html>
  <p>Olá, time,</p>

    <p>Resumo dos fatos relevantes do período <b>04–05/08/2026</b>: 4 comunicados — <b>1 positivo e 3 neutros</b>. Relatório completo em anexo.</p>

    <p><b>PMLL11 · Pátria Malls FII — Positivo, condicionado à conclusão</b><br>
    Intenção de compra de 10% do Shopping Jardim Sul por R$ 67,0 milhões. A gestora estima yield médio de 11,6% a.a. nos primeiros dois anos e 9,1% a.a. na estabilidade.<br>
    → <i>Expansão da exposição a shopping consolidado, com parcela do preço vinculada à subscrição de cotas pelo vendedor.</i></p>

    <p><b>SPX SYN Desenvolvimento de Galpões Logísticos Feeder Bradesco FII — Neutro</b><br>
    4ª chamada de capital da 1ª emissão: 831.171 cotas a R$ 10,00, totalizando R$ 8,31 milhões. Liquidação prevista para 14/08/2026.<br>
    → <i>Integralização de compromissos de capital já assumidos, sem anúncio de uma nova oferta.</i></p>

    <p><b>SPX Desenvolvimento de Galpões Logísticos Feeder Institucional FII — Neutro</b><br>
    Chamadas de capital para cotas já subscritas: R$ 20,84 milhões na 1ª emissão e R$ 12,51 milhões na 2ª emissão, com liquidação em 14/08/2026.<br>
    → <i>Conversão de capital subscrito em recursos disponíveis ao fundo; não representa captação incremental.</i></p>

    <p><b>CPLG11 · Capitânia Logística FII — Neutro</b><br>
    Aprovação da 6ª emissão de até R$ 300,0 milhões, correspondentes a até 27,7 milhões de cotas, ao preço de subscrição de R$ 10,84 por cota. Captação mínima de R$ 50,0 milhões, com direito de preferência.<br>
    → <i>A emissão amplia o poder de investimento, mas traz risco de diluição para cotistas que não acompanhem a oferta.</i></p>

    <p>Att,<br>

    Lucas Cavalcante.</p>
  </body>
</html>
"""
    msg.attach(MIMEText(body, 'html'))

    # Anexar o arquivo PDF
    try:
        with open(pdf_path, "rb") as attachment:
            part = MIMEBase("application", "octet-stream")
            part.set_payload(attachment.read())

        # Codificar o arquivo em Base64
        encoders.encode_base64(part)

        # Adicionar cabeçalhos para o anexo
        part.add_header(
            "Content-Disposition",
            f"attachment; filename= {os.path.basename(pdf_path)}",  # Nome do arquivo
        )

        # Anexar o arquivo PDF à mensagem
        msg.attach(part)

    except Exception as e:
        print(f"Falha ao anexar o arquivo: {e}")
        return

    # Conectar ao servidor SMTP do Gmail e enviar o e-mail
    try:
        with smtplib.SMTP_SSL('smtp.gmail.com', 465) as smtp_server:
            smtp_server.login(sender_email, sender_password)
            smtp_server.sendmail(sender_email, recipient_email, msg.as_string())

        print(f"E-mail enviado para {', '.join(recipient_email)}")

    except Exception as e:
        print(f"Falha ao enviar e-mail: {e}")
```
