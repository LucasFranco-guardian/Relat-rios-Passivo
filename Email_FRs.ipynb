```python
import smtplib
import os
from email.mime.multipart import MIMEMultipart
from email.mime.text import MIMEText
from email.mime.base import MIMEBase
from email import encoders
from datetime import datetime
```

```python
def send_email_with_pdf_FRs(pdf_path):
    # Configurações do remetente e destinatários
    sender_email = 'lc@guardian-asset.com'  # Substitua pelo seu e-mail
    sender_password = os.getenv("senha_gmail")  # Substitua pela sua senha ou senha de aplicativo
    recipient_email = ['lc@guardian-asset.com','pmk@guardian-asset.com', 'msf@guardian-asset.com', 'cm@guardian-asset.com', 'lb@guardian-asset.com', 'ja@guardian-asset.com', 'gsa@guardian-asset.com',
                       'gvm@guardian-asset.com', 'rt@guardian-asset.com', 'acl@guardian-asset.com', 'lt@guardian-asset.com', 'tfa@guardian-asset.com']  # Lista de destinatários

    # Criando a mensagem
    msg = MIMEMultipart()
    msg['From'] = sender_email
    msg['To'] = ", ".join(recipient_email)
    msg['Subject'] = "Fatos Relevantes — Guardian Gestora | 05–06/08/2026"

    # Corpo do e-mail (HTML para preservar negrito, itálico e as setas)
    body = """\
<html>
  <p>Olá, time,</p>

    <p>Resumo dos fatos relevantes do período <b>05–06/08/2026</b>: 7 comunicados — <b>5 neutros, 1 positivo e 1 negativo</b>. Relatório completo em anexo.</p>

    <p><b>TRXF11 · TRX Real Estate FII — Neutro</b><br>
    Aprovada a 13ª emissão de cotas, com preço de R$ 94,25 e captação de até R$ 5,0 bilhões.<br>
    → <i>Reforça a capacidade de investimento em desenvolvimento e sale and leaseback, com potencial diluição para quem não acompanhar a oferta.</i></p>

    <p><b>TRXF11 · TRX Real Estate FII — Neutro</b><br>
    MoU não vinculante para adquirir 70% do LOG Recife II por R$ 210,0 milhões.<br>
    → <i>Entrada potencial em ativo logístico locado, ainda condicionada a diligências, oferta de cotas e aprovação concorrencial.</i></p>

    <p><b>TRXF11 · TRX Real Estate FII — Neutro</b><br>
    MoU para aquisição de 5 imóveis por R$ 2,13 bilhões, com ABL de 251.164 m² e YoC de 10,40% a/a.<br>
    → <i>Expansão relevante da carteira logística e corporativa, dependente da estrutura de financiamento e de aprovação concorrencial.</i></p>

    <p><b>HGBS11 · Hedge Brasil Shopping FII — Positivo</b><br>
    Compromisso para adquirir 75% do Shopping Jaraguá Araraquara por R$ 216,3 milhões.<br>
    → <i>Se concluída, a operação elevará a participação do fundo a 100% no ativo; ainda há condições precedentes e aprovação concorrencial.</i></p>

    <p><b>KNPR11 · Kinea Premium Properties FII — Negativo</b><br>
    Alerta de que a cota negociava com ágio de 44%, a R$ 40,00, versus valor patrimonial de R$ 27,84.<br>
    → <i>O fundo está em desinvestimento; amortizações futuras dependem da venda de ativos e não acompanham necessariamente a cotação de bolsa.</i></p>

    <p><b>BTHF11 · BTG Pactual Real Estate Hedge Fund FII — Neutro</b><br>
    Aquisição de 76% do Shopping Pátio Cianê por R$ 220,4 milhões, incluindo assunção de dívida de R$ 66,9 milhões a IPCA + 7,65% a/a.<br>
    → <i>A aquisição adiciona renda imobiliária potencial, mas também dívida e compromissos corrigidos por IPCA.</i></p>

    <p><b>INDE11 · Inter Desenvolvimento FII — Neutro</b><br>
    Com o adiamento de projetos, houve renúncia das taxas de gestão e consultoria e taxa de administração de R$ 3 mil no período indicado.<br>
    → <i>A redução de despesas é favorável, mas decorre do adiamento dos lançamentos de projetos.</i></p>

    <p>Att,<br>

    Lucas Cavalcante.</p>
  </body>
</html>
"""
    msg.attach(MIMEText(body, 'html'))

    # Anexar o arquivo PDF
    try:
        with open(pdf_path, "rb") as attachment:
            part = MIMEBase("application", "octet-stream")
            part.set_payload(attachment.read())

        # Codificar o arquivo em Base64
        encoders.encode_base64(part)

        # Adicionar cabeçalhos para o anexo
        part.add_header(
            "Content-Disposition",
            f"attachment; filename= {os.path.basename(pdf_path)}",  # Nome do arquivo
        )

        # Anexar o arquivo PDF à mensagem
        msg.attach(part)

    except Exception as e:
        print(f"Falha ao anexar o arquivo: {e}")
        return

    # Conectar ao servidor SMTP do Gmail e enviar o e-mail
    try:
        with smtplib.SMTP_SSL('smtp.gmail.com', 465) as smtp_server:
            smtp_server.login(sender_email, sender_password)
            smtp_server.sendmail(sender_email, recipient_email, msg.as_string())

        print(f"E-mail enviado para {', '.join(recipient_email)}")

    except Exception as e:
        print(f"Falha ao enviar e-mail: {e}")
```
