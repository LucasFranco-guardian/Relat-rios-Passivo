```python
import smtplib
import os
from email.mime.multipart import MIMEMultipart
from email.mime.text import MIMEText
from email.mime.base import MIMEBase
from email import encoders
from datetime import datetime
```

```python
def send_email_with_pdf_FRs(pdf_path):
    # Configurações do remetente e destinatários
    sender_email = 'lc@guardian-asset.com'  # Substitua pelo seu e-mail
    sender_password = os.getenv("senha_gmail")  # Substitua pela sua senha ou senha de aplicativo
    recipient_email = ['lc@guardian-asset.com','pmk@guardian-asset.com', 'msf@guardian-asset.com', 'cm@guardian-asset.com', 'lb@guardian-asset.com', 'ja@guardian-asset.com', 'gsa@guardian-asset.com',
                       'gvm@guardian-asset.com', 'rt@guardian-asset.com', 'acl@guardian-asset.com', 'lt@guardian-asset.com', 'tfa@guardian-asset.com']  # Lista de destinatários

    # Criando a mensagem
    msg = MIMEMultipart()
    msg['From'] = sender_email
    msg['To'] = ", ".join(recipient_email)
    msg['Subject'] = "Fatos Relevantes — Guardian Gestora | 09–10/09/2026"

    # Corpo do e-mail (HTML para preservar negrito, itálico e as setas)
    body = """\
<html>
        <p>Olá, time,</p>

    <p>Resumo dos fatos relevantes do período <b>09/09/2026 a 10/09/2026</b>: 4 comunicados — <b>2 positivos e 2 neutros</b>. Relatório completo em anexo.</p>

    <p><b>TRXF11 · TRX Real Estate FII — Positivo</b><br>
    MoU para alienação do imóvel na Av. Ricardo Jafet, 1550 (São Paulo/SP), locado à Dasa, por R$ 77.000.000,00 à vista na escritura, prevista em até 45 dias corridos. Lucro estimado de aprox. R$ 3,6 milhões (R$ 0,06 por cota) e TIR aprox. de 16,28%.<br>
    → <i>Desinvestimento imobiliário com ganho de capital relevante e taxa de retorno atrativa, gerando potencial incremento na distribuição aos cotistas.</i></p>

    <p><b>SNEL11 · Suno Energias Limpas FII IS — Positivo</b><br>
    Retratação e cancelamento de 14.734 cotas condicionadas da 5ª emissão. Consolidado final: 116.374.627 novas cotas integralizadas, somando R$ 968.236.896,64 (sem taxa) e R$ 1.006.640.523,55 (com taxa de distribuição).<br>
    → <i>Conclusão bem-sucedida de captação superior a R$ 1 bilhão, com efeito marginal da retratação de cotas.</i></p>

    <p><b>FTRR11 · Fator Recebíveis Imobiliários FII — Neutro</b><br>
    Divulgação do Prospecto Definitivo, Lâmina e Comunicado de 5ª Modificação da 1ª emissão: até 40.000.000 cotas a R$ 10,00, montante inicial de até R$ 400.000.000,00 (lote adicional de até 25% / R$ 100.000.000,00), para o público em geral. A 5ª modificação incluiu data de integralização com compensação de créditos e período de desistência de 10/09/2026 a 16/09/2026 (até 16h).<br>
    → <i>Ajustes operacionais de cronograma e termos de integralização de créditos da captação inicial do fundo.</i></p>

    <p><b>MCRE11 · Mauá Capital Real Estate FII — Neutro</b><br>
    Encerrado o direito de preferência da 4ª emissão: 19.652 cotas subscritas e integralizadas (R$ 199.860,84), restando 9.813.190 cotas (R$ 99.800.142,30) para colocação no período de coleta de intenções de investimento.<br>
    → <i>Conclusão do direito de preferência com alocação concentrada na fase de sobras/oferta pública.</i></p>

    <p>Att,<br>

    Lucas Cavalcante.</p>
  </body>
</html>
"""
    msg.attach(MIMEText(body, 'html'))

    # Anexar o arquivo PDF
    try:
        with open(pdf_path, "rb") as attachment:
            part = MIMEBase("application", "octet-stream")
            part.set_payload(attachment.read())

        # Codificar o arquivo em Base64
        encoders.encode_base64(part)

        # Adicionar cabeçalhos para o anexo
        part.add_header(
            "Content-Disposition",
            f"attachment; filename= {os.path.basename(pdf_path)}",  # Nome do arquivo
        )

        # Anexar o arquivo PDF à mensagem
        msg.attach(part)

    except Exception as e:
        print(f"Falha ao anexar o arquivo: {e}")
        return

    # Conectar ao servidor SMTP do Gmail e enviar o e-mail
    try:
        with smtplib.SMTP_SSL('smtp.gmail.com', 465) as smtp_server:
            smtp_server.login(sender_email, sender_password)
            smtp_server.sendmail(sender_email, recipient_email, msg.as_string())

        print(f"E-mail enviado para {', '.join(recipient_email)}")

    except Exception as e:
        print(f"Falha ao enviar e-mail: {e}")
```
