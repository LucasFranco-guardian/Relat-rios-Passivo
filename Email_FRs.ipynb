```python
import smtplib
import os
from email.mime.multipart import MIMEMultipart
from email.mime.text import MIMEText
from email.mime.base import MIMEBase
from email import encoders
from datetime import datetime
```

```python
def send_email_with_pdf_FRs(pdf_path):
    # Configurações do remetente e destinatários
    sender_email = 'lc@guardian-asset.com'  # Substitua pelo seu e-mail
    sender_password = os.getenv("senha_gmail")  # Substitua pela sua senha ou senha de aplicativo
    recipient_email = ['lc@guardian-asset.com','pmk@guardian-asset.com', 'msf@guardian-asset.com', 'cm@guardian-asset.com', 'lb@guardian-asset.com', 'ja@guardian-asset.com', 'gsa@guardian-asset.com',
                       'gvm@guardian-asset.com', 'rt@guardian-asset.com', 'acl@guardian-asset.com', 'lt@guardian-asset.com', 'tfa@guardian-asset.com']  # Lista de destinatários

    # Criando a mensagem
    msg = MIMEMultipart()
    msg['From'] = sender_email
    msg['To'] = ", ".join(recipient_email)
    msg['Subject'] = "Fatos Relevantes — Guardian Gestora | 10–11/09/2026"

    # Corpo do e-mail (HTML para preservar negrito, itálico e as setas)
    body = """\
<html>
    <p>Olá, time,</p>

    <p>Resumo dos fatos relevantes do período <b>10–11/09/2026</b>: 7 comunicados — <b>3 positivos e 4 neutros</b>. Relatório completo em anexo.</p>

    <p><b>HGBS11 · Hedge Brasil Shopping FII — Positivo</b><br>
    Conclusão da alienação de 19% do Shopping Jardim Sul por R$ 135.443.939,44 (cap rate de 7,5%, 23,6% acima do laudo). Recebimento de R$ 72,81 MM em cotas de PMLL11, R$ 12,53 MM à vista e duas parcelas de R$ 25,05 MM corrigidas pelo IPCA; lucro total de R$ 22.219.695,38 (R$ 0,154/cota).<br>
    → <i>Desinvestimento acima do valor de laudo com expressivo ganho de capital e reforço do guidance de distribuição.</i></p>

    <p><b>PMLL11 · Pátria Malls FII — Positivo</b><br>
    Conclusão da aquisição de 29% do Shopping Jardim Sul (28.741 m² de ABL total) por R$ 204.238.905,36 — R$ 95.419.250,00 em dinheiro e R$ 108.819.655,36 via compensação de créditos na 7ª emissão. Yield médio estimado de 11,7% a.a. nos 2 primeiros anos e aprox. 9,9% a.a. na estabilidade.<br>
    → <i>Expansão de portfólio de shopping centers com yield inicial elevado e estrutura de pagamento favorável.</i></p>

    <p><b>RBRP11 · Pátria Properties FII — Positivo</b><br>
    Assinatura de CVC para alienação total do Edifício Jacks Rabinovich para a Fortinet. Pela participação de 60% do fundo (2.865 m² de ABL), o valor recebido será de R$ 100.500.000,00. A operação gera lucro por cota de R$ 2,83, TIR de 17,8% a.a., ganho de capital de R$ 34,4 MM e MOIC de 1,67x.<br>
    → <i>Reciclagem de portfólio corporativo com retorno de alta rentabilidade (TIR de 17,8% a.a.) e forte geração de caixa e lucro.</i></p>

    <p><b>CYHF11 · Cyrela Hedge Fund FII — Neutro</b><br>
    Aprovação da 2ª emissão de cotas no volume inicial de R$ 350.000.002,80 (35.714.286 cotas a R$ 9,80), com lote adicional de até 25% (até R$ 437.499.998,60) e montante mínimo de R$ 50.000.001,80. Fator de proporção para preferência de 1,69896449338 (data-base 24/07/2026).<br>
    → <i>Aprovação de nova captação expressiva para expansão do fundo de papel/crédito imobiliário.</i></p>

    <p><b>TRXF11 · TRX Real Estate FII — Neutro</b><br>
    Encerramento do período de preferência e sobras da 13ª emissão (425.272 cotas na preferência e 948 nas sobras). Início da colocação de 52.624.178 cotas remanescentes junto a investidores profissionais, com prazo de até 180 dias.<br>
    → <i>Avanço das fases da 13ª emissão com início da distribuição de cotas remanescentes a investidores profissionais.</i></p>

    <p><b>TRX Logístico FII — Neutro</b><br>
    Início da oferta pública primária da 1ª emissão no montante inicial de R$ 1.460.000.000,00 (14.600.000 cotas a R$ 100,00), sendo R$ 930 MM em Cotas Seniores e R$ 530 MM em Cotas Subordinadas. Distribuição parcial mínima de R$ 1.000.000,00 e 1ª liquidação tentativa em 15/09/2026.<br>
    → <i>Oferta primária inaugural de grande porte para investidores profissionais em estrutura multi-classe.</i></p>

    <p><b>Bresco BTS FII — Neutro</b><br>
    Divulgação da lâmina da 1ª emissão no montante inicial de R$ 170.000.000,00 (1.700.000 cotas a R$ 100,00), lote adicional de até 25% e aplicação mínima de R$ 50.000,00. Prazo de duração de 5 anos e 1ª liquidação estimada para 30/10/2026.<br>
    → <i>Estruturação de captação primária em estratégia sob medida (Built-to-Suit) com prazo de 5 anos.</i></p>

    <p>Att,<br>

    Lucas Cavalcante.</p>
  </body>
</html>
"""
    msg.attach(MIMEText(body, 'html'))

    # Anexar o arquivo PDF
    try:
        with open(pdf_path, "rb") as attachment:
            part = MIMEBase("application", "octet-stream")
            part.set_payload(attachment.read())

        # Codificar o arquivo em Base64
        encoders.encode_base64(part)

        # Adicionar cabeçalhos para o anexo
        part.add_header(
            "Content-Disposition",
            f"attachment; filename= {os.path.basename(pdf_path)}",  # Nome do arquivo
        )

        # Anexar o arquivo PDF à mensagem
        msg.attach(part)

    except Exception as e:
        print(f"Falha ao anexar o arquivo: {e}")
        return

    # Conectar ao servidor SMTP do Gmail e enviar o e-mail
    try:
        with smtplib.SMTP_SSL('smtp.gmail.com', 465) as smtp_server:
            smtp_server.login(sender_email, sender_password)
            smtp_server.sendmail(sender_email, recipient_email, msg.as_string())

        print(f"E-mail enviado para {', '.join(recipient_email)}")

    except Exception as e:
        print(f"Falha ao enviar e-mail: {e}")
```
