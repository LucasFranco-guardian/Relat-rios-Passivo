```python
import smtplib
import os
from email.mime.multipart import MIMEMultipart
from email.mime.text import MIMEText
from email.mime.base import MIMEBase
from email import encoders
from datetime import datetime
```

```python
def send_email_with_pdf_FRs(pdf_path):
    # Configurações do remetente e destinatários
    sender_email = 'lc@guardian-asset.com'  # Substitua pelo seu e-mail
    sender_password = os.getenv("senha_gmail")  # Substitua pela sua senha ou senha de aplicativo
    recipient_email = ['lc@guardian-asset.com','pmk@guardian-asset.com', 'msf@guardian-asset.com', 'cm@guardian-asset.com', 'lb@guardian-asset.com', 'ja@guardian-asset.com', 'gsa@guardian-asset.com',
                       'gvm@guardian-asset.com', 'rt@guardian-asset.com', 'acl@guardian-asset.com', 'lt@guardian-asset.com', 'tfa@guardian-asset.com', 'mer@guardian-asset.com']  # Lista de destinatários

    # Criando a mensagem
    msg = MIMEMultipart()
    msg['From'] = sender_email
    msg['To'] = ", ".join(recipient_email)
    msg['Subject'] = "Fatos Relevantes — Guardian Gestora | 23–26/10/2026"

    # Corpo do e-mail (HTML para preservar negrito, itálico e as setas)
    body = """\
<html>
    <p>Olá, time,</p>

    <p>Resumo dos fatos relevantes do período <b>22–23/10/2026</b>: 5 comunicados — <b>2 positivos, 2 neutros e 1 negativo</b>. Relatório completo em anexo.</p>

    <p><b>HGLG11 · Pátria Log FII — Neutro</b><br>
    Comunicado sobre a incorporação do PATL11: detentores de frações de cotas recebem R$ 1,17 por cota em cada mês (competências de agosto e setembro de 2026), totalizando R$ 2,34 por cota.<br>
    → <i>Procedimento operacional de liquidação e acerto financeiro relativo à incorporação e amortização do PATL11, distribuindo os proventos apurados aos detentores das frações.</i></p>

    <p><b>HGLG11 · Pátria Log FII — Positivo</b><br>
    Aditamento formaliza a aquisição de 100% das quotas da sociedade dona do Pátio GRU (estacionamento de carretas no Parque Logístico Guarulhos) por R$ 9.975.444,57 à vista, a um cap rate de 9,1%.<br>
    → <i>Expansão de portfólio logístico com ativo diretamente integrado ao complexo já adquirido, com cap rate atrativo e geração de renda imediata para a carteira.</i></p>

    <p><b>PSEC11 · Pátria Securities FII — Neutro</b><br>
    Assembleia aprovou a consolidação do patrimônio do PSEC11 pelo RBRX11, com migração dos cotistas e liquidação do fundo. Os ativos serão alienados pelo valor patrimonial, com pagamento em cotas do RBRX11 e/ou caixa.<br>
    → <i>Reorganização societária para ganho de escala e liquidez no RBRX11, com transferência de ativos a valor patrimonial contábil e encerramento de veículo menor.</i></p>

    <p><b>TRXF11 · TRX Real Estate FII — Positivo</b><br>
    Contratação da Suno S/A como formador de mercado das cotas na B3, por 12 meses.<br>
    → <i>Ação voltada a dar maior liquidez e profundidade de livro às cotas no mercado secundário da B3, reduzindo o spread entre ordens de compra e venda.</i></p>

    <p><b>Sem ticker · LCP Renda Imobiliária Master FII — Negativo</b><br>
    Inadimplência na Operação Poehma (CRIs da Habitasec), equivalente a 5,4% do PL, sem pagamento de juros e amortização desde 15/06/2026. Obras concluídas, 66,69% comercializado e garantias de 161% do saldo. Por enquanto, sem remarcação nem impacto na cota sênior.<br>
    → <i>Evento de risco de crédito relevante (5,4% do PL), temporariamente amortecido pela estrutura de subordinação (23%) e pelas garantias reais em fase de acompanhamento.</i></p>

    Lucas Cavalcante.</p>
  </body>
</html>
"""
    msg.attach(MIMEText(body, 'html'))

    # Anexar o arquivo PDF
    try:
        with open(pdf_path, "rb") as attachment:
            part = MIMEBase("application", "octet-stream")
            part.set_payload(attachment.read())

        # Codificar o arquivo em Base64
        encoders.encode_base64(part)

        # Adicionar cabeçalhos para o anexo
        part.add_header(
            "Content-Disposition",
            f"attachment; filename= {os.path.basename(pdf_path)}",  # Nome do arquivo
        )

        # Anexar o arquivo PDF à mensagem
        msg.attach(part)

    except Exception as e:
        print(f"Falha ao anexar o arquivo: {e}")
        return

    # Conectar ao servidor SMTP do Gmail e enviar o e-mail
    try:
        with smtplib.SMTP_SSL('smtp.gmail.com', 465) as smtp_server:
            smtp_server.login(sender_email, sender_password)
            smtp_server.sendmail(sender_email, recipient_email, msg.as_string())

        print(f"E-mail enviado para {', '.join(recipient_email)}")

    except Exception as e:
        print(f"Falha ao enviar e-mail: {e}")
```
