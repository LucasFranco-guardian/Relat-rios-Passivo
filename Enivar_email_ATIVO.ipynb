```python
# ============================================================
# 3) FUNÇÕES DE ENVIO — uma implementação, usada pelos dois relatórios
# ============================================================
REMETENTE = "lc@guardian-asset.com"

DESTINATARIOS = [
    "lc@guardian-asset.com", "pmk@guardian-asset.com", "msf@guardian-asset.com",
    "cm@guardian-asset.com", "lb@guardian-asset.com", "ja@guardian-asset.com",
    "gsa@guardian-asset.com", "gvm@guardian-asset.com", "rt@guardian-asset.com",
    "acl@guardian-asset.com", "lt@guardian-asset.com", "tfa@guardian-asset.com", "ffs@guardian-asset.com",
]

# Fecha todo e-mail, nos dois relatórios (anexada por _html_email).
ASSINATURA = """<p>Att,<br>

Lucas Cavalcante.</p>"""


def _html_email(corpo):
    """Embrulha o miolo HTML no mesmo estilo usado no relatório PASSIVO."""
    return (
        '<html><body style="font-family: Arial, sans-serif; font-size: 14px; '
        'color: #222222; line-height: 1.55;">'
        f"{corpo.strip()}"
        f"{ASSINATURA}"
        "</body></html>"
    )


def _enviar_email(pdf_path, corpo, assunto):
    msg = MIMEMultipart()
    msg["From"] = REMETENTE
    msg["To"] = ", ".join(DESTINATARIOS)
    msg["Subject"] = assunto
    msg.attach(MIMEText(_html_email(corpo), "html"))

    # Anexar o PDF
    try:
        with open(pdf_path, "rb") as attachment:
            part = MIMEBase("application", "octet-stream")
            part.set_payload(attachment.read())
        encoders.encode_base64(part)
        part.add_header(
            "Content-Disposition",
            f"attachment; filename= {os.path.basename(pdf_path)}",
        )
        msg.attach(part)
    except Exception as e:
        print(f"❌ Falha ao anexar o arquivo: {e}")
        return

    # Enviar via SMTP do Gmail (credencial via .env: senha_gmail)
    try:
        with smtplib.SMTP_SSL("smtp.gmail.com", 465) as smtp_server:
            smtp_server.login(REMETENTE, os.getenv("senha_gmail"))
            smtp_server.sendmail(REMETENTE, DESTINATARIOS, msg.as_string())
        print(f"✅ E-mail enviado — {assunto}")
    except Exception as e:
        print(f"❌ Falha ao enviar e-mail: {e}")


def _enviar_whatsapp(texto, pdf_path):
    """Envia o texto e, em seguida, o PDF ao grupo do ATIVO."""
    if not GRUPO_ATIVO_PHONE:
        raise ValueError("GRUPO_ATIVO_PHONE não definido no .env.")
    enviar_texto_whatsapp(texto, phone=GRUPO_ATIVO_PHONE)
    enviar_documento_whatsapp(pdf_path, phone=GRUPO_ATIVO_PHONE)


def enviar_ativo(tipo, pdf_path, corpo, texto_whats, data, email=True, whatsapp=True):
    """tipo: 'TIJOLO' ou 'PAPEL'. Assunto: RELATÓRIO DE ATIVO {tipo} DIA {data}."""
    if email:
        _enviar_email(pdf_path, corpo, f"RELATÓRIO DE ATIVO {tipo} DIA {data}")
    if whatsapp:
        _enviar_whatsapp(texto_whats, pdf_path)
```
