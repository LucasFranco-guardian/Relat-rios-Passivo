```python
import smtplib
import os
from email.mime.multipart import MIMEMultipart
from email.mime.text import MIMEText
from email.mime.base import MIMEBase
from email import encoders
from datetime import datetime
from dotenv import load_dotenv

from resumo_ativo import gerar_resumo, resumo_para_html, enviar_whatsapp_grupo

load_dotenv()
```

```python
# ============================================================
# FUNÇÕES DE E-MAIL (corpo HTML com o resumo + PDF em anexo)
# Credencial via .env (senha_gmail) — sem senha hardcoded.
# ============================================================
DESTINATARIOS = [
    'lc@guardian-asset.com', 'pmk@guardian-asset.com', 'msf@guardian-asset.com',
    'cm@guardian-asset.com', 'lb@guardian-asset.com', 'ja@guardian-asset.com',
    'gsa@guardian-asset.com', 'gvm@guardian-asset.com', 'rt@guardian-asset.com',
    'acl@guardian-asset.com', 'lt@guardian-asset.com',
]


def _enviar_email(pdf_path, resumo, assunto, saudacao):
    sender_email = 'lc@guardian-asset.com'
    sender_password = os.getenv("senha_gmail")

    msg = MIMEMultipart()
    msg['From'] = sender_email
    msg['To'] = ", ".join(DESTINATARIOS)
    msg['Subject'] = assunto

    # Corpo em HTML, com o resumo formatado (negrito, separadores, emojis)
    msg.attach(MIMEText(resumo_para_html(resumo, saudacao=saudacao), 'html'))

    # Anexar o PDF
    try:
        with open(pdf_path, "rb") as attachment:
            part = MIMEBase("application", "octet-stream")
            part.set_payload(attachment.read())
        encoders.encode_base64(part)
        part.add_header(
            "Content-Disposition",
            f"attachment; filename= {os.path.basename(pdf_path)}",
        )
        msg.attach(part)
    except Exception as e:
        print(f"Falha ao anexar o arquivo: {e}")
        return

    # Enviar via SMTP do Gmail
    try:
        with smtplib.SMTP_SSL('smtp.gmail.com', 465) as smtp_server:
            smtp_server.login(sender_email, sender_password)
            smtp_server.sendmail(sender_email, DESTINATARIOS, msg.as_string())
        print(f"E-mail enviado para {', '.join(DESTINATARIOS)}")
    except Exception as e:
        print(f"Falha ao enviar e-mail: {e}")


def send_email_with_pdf_TIJOLO(pdf_path, resumo, data):
    _enviar_email(
        pdf_path, resumo,
        assunto=f"RELATÓRIO DE ATIVO TIJOLO DIA {data}",
        saudacao=f"Prezados, segue o relatório do mercado dos fundos de TIJOLO do dia {data}. Resumo abaixo:",
    )


def send_email_with_pdf_PAPEL(pdf_path, resumo, data):
    _enviar_email(
        pdf_path, resumo,
        assunto=f"RELATÓRIO DE ATIVO PAPEL DIA {data}",
        saudacao=f"Prezados, segue o relatório do mercado dos fundos de PAPEL do dia {data}. Resumo abaixo:",
    )
```
